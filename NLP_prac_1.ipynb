```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.naive_bayes import MultinomialNB
import html
import re
```

```python
df = pd.read_parquet("comments_data_13.parquet")
df
```

```python
df = df[df["channel_id"] != df["author_channel_id"]]
# df = df[df["like_count"] == 1]
df
```

```python
# 1. First, apply your hard constraints (drop HTML, author comments, and short spam)
df = df[df['text_display'].str.split().str.len() >= 10]

# 2. Create a 'Quality Score': Word Count gives text depth, plus a small boost for likes
# This ensures long, thoughtful questions with 0 likes beat out a 2-word comment with 50 likes
df['quality_score'] = df['text_display'].str.split().str.len() + (df['like_count'] * 2)

# 3. Sort by video and quality, then slice the top 20 comments for every single video
df_top_comments = (
    df.sort_values(by=['video_id', 'quality_score'], ascending=[False, True])
    .groupby('video_id')
    .head(20)  # Keep up to the 20 best comments per video
)
```

```python
df
```

```python
df = df.drop_duplicates(subset = ["text_display"])
df
```

```python
df["text_display"] = df["text_display"].replace(r"https?://\S+|www\.\S+", "", regex = True).iloc[:]
df["text_display"] = df["text_display"].replace(r"<[^<>]*>", "", regex = True)
df["text_display"] = df["text_display"].replace(r"['’][sS]\b","",regex = True)
df["text_display"] = df["text_display"].replace(r"\b\d{1,2}:\d{2}(:\d{2})?\b", "", regex = True)
df["text_display"] = df["text_display"].replace(r"\s+"," ", regex = True).str.strip()
df["text_display"] = df["text_display"].replace(r"(.)\1{2,}", r"\1\1", regex = True)
df["text_display"] = df["text_display"].apply(html.unescape)
```

```python
latin_count = df["text_display"].str.count(r'[a-zA-Z0-9\s.,?!’\'"]')
total_count = df["text_display"].str.len()
# Set 0 total count to 1 to avoid Division by Zero errors
ratio = latin_count / total_count.replace(0, 1)
df = df[ratio > 0.9]
```

```python
df["text_display"]
```

```python
df = df.groupby('video_id')['text_display'].apply(lambda x: ' '.join(x)).reset_index()
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
)

# Convert the Series to a list of strings directly
comments = df["text_display"].tolist()

embeddings = model.encode(
    comments,
    batch_size=16,
    normalize_embeddings=True
)

print(embeddings)
```

```python

model = SentenceTransformer(
    "google/muril-base-cased"
)

# Convert the Series to a list of strings directly
comments = df["text_display"].tolist()

embeddings = model.encode(
    comments,
    batch_size=16,
    normalize_embeddings=True
)

print(embeddings)
```
